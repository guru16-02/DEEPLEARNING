# Simple RNN - Next Word Prediction

This notebook builds a simple RNN model to predict the next word from a given sequence.

## 1. Import Libraries

```python
import numpy as np
import tensorflow as tf

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.utils import to_categorical
```

## 2. Create the Text Dataset

```python
text = """
I love machine learning
I love deep learning
I love artificial intelligence
machine learning is interesting
deep learning is powerful
artificial intelligence is amazing
I study machine learning
I study artificial intelligence
"""
```

## 3. Tokenization

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

word_index = tokenizer.word_index
total_words = len(word_index) + 1

print("Word Index:")
print(word_index)
print("\nTotal Words:", total_words)
```

## 4. Create N-gram Sequences

```python
input_sequences = []

for line in text.strip().split("\n"):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i + 1]
        input_sequences.append(n_gram_sequence)

print("Input Sequences:")
print(input_sequences)
```

## 5. Find Maximum Sequence Length

```python
max_sequence_len = max([len(sequence) for sequence in input_sequences])
print("Maximum Sequence Length:", max_sequence_len)
```

## 6. Separate X and y

```python
X = []
y = []

for sequence in input_sequences:
    X.append(sequence[:-1])
    y.append(sequence[-1])

print("X:")
print(X)

print("\ny:")
print(y)
```

## 7. Padding

```python
X = pad_sequences(
    X,
    maxlen=max_sequence_len - 1,
    padding='pre'
)

print("Padded X:")
print(X)
```

## 8. Convert y to Categorical Format

```python
y = to_categorical(y, num_classes=total_words)

print("Shape of X:", X.shape)
print("Shape of y:", y.shape)
```

## 9. Build the Simple RNN Model

```python
model = Sequential()

model.add(Embedding(
    input_dim=total_words,
    output_dim=10
))

model.add(SimpleRNN(50))

model.add(Dense(
    total_words,
    activation='softmax'
))
```

## 10. Compile the Model

```python
model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)
```

## 11. Display Model Structure

```python
model.summary()
```

## 12. Train the Model

```python
history = model.fit(
    X,
    y,
    epochs=200,
    verbose=1
)
```

## 13. Predict the Next Word

```python
def predict_next_word(seed_text):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]

    token_list = pad_sequences(
        [token_list],
        maxlen=max_sequence_len - 1,
        padding='pre'
    )

    predicted = model.predict(token_list, verbose=0)
    predicted_word_index = np.argmax(predicted)

    for word, index in tokenizer.word_index.items():
        if index == predicted_word_index:
            return word

    return None
```

## 14. Test the Model

```python
seed_text = "I love"

next_word = predict_next_word(seed_text)

print("Input:", seed_text)
print("Predicted Next Word:", next_word)
```

## Project Flow

Text → Tokenization → N-gram Sequences → X and y → Padding → Embedding → Simple RNN → Dense + Softmax → Next Word Prediction
