```python
# ============================================
# PART A: LOADING AND CONFIGURING PRE-TRAINED LLM
# Name: Guru prasad N
# Roll Number: 24BAD029
# ============================================

# Install required libraries


# Import libraries
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# Select pre-trained model
model_name = "microsoft/DialoGPT-small"

# Load tokenizer
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Load pre-trained model
model = AutoModelForCausalLM.from_pretrained(model_name)

# Select device
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model.to(device)

print("Model loaded successfully!")
print("Model Name:", model_name)
print("Device:", device)

# Sample user prompt
prompt = "Hello! How are you?"

# Tokenize input
input_ids = tokenizer.encode(
    prompt + tokenizer.eos_token,
    return_tensors='pt'
).to(device)

print("\nTokenized Input:")
print(input_ids)

# Generate response
with torch.no_grad():

    output = model.generate(
        input_ids,
        max_new_tokens=50,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

# Decode response
response = tokenizer.decode(
    output[:, input_ids.shape[-1]:][0],
    skip_special_tokens=True
)

print("\nUser Prompt:", prompt)
print("Generated Response:", response)

print("\nPART A COMPLETED SUCCESSFULLY")

```

```python


# Conversation history
chat_history_ids = None

# Chatbot response function
def chatbot_response(user_input):

    global chat_history_ids

    # Tokenize user input
    new_input_ids = tokenizer.encode(
        user_input + tokenizer.eos_token,
        return_tensors='pt'
    ).to(device)

    # Combine with conversation history
    if chat_history_ids is not None:

        bot_input_ids = torch.cat(
            [chat_history_ids, new_input_ids],
            dim=-1
        )

    else:

        bot_input_ids = new_input_ids

    # Generate chatbot response
    with torch.no_grad():

        chat_history_ids = model.generate(
            bot_input_ids,
            max_new_tokens=60,
            pad_token_id=tokenizer.eos_token_id,
            do_sample=True,
            temperature=0.7,
            top_p=0.9
        )

    # Decode generated response
    response = tokenizer.decode(
        chat_history_ids[:, bot_input_ids.shape[-1]:][0],
        skip_special_tokens=True
    )

    return response


# Test chatbot with different queries
queries = [
    "Hello!",
    "What is artificial intelligence?",
    "What is machine learning?",
    "Tell me about Python programming."
]

for query in queries:

    print("\nUser:", query)

    answer = chatbot_response(query)

    print("Chatbot:", answer)

print("\nPART B COMPLETED SUCCESSFULLY")
```
