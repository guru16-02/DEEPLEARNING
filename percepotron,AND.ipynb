```python
input=[
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]
target=[0,0,0,1]
```

```python

w1=0
w2=0
bias=0
learning_rate=0.1
```

```python
for epoch in range(10):
    print("epoch:",epoch+1)
    for i in range(len(input)):
        x1,x2 = input[i]

        act= w1*x1+w2*x2+bias

        if act >= 0:
            output=1
        else:
            output=0
        
        error = target[i] - output   
        
        #update the weights
        w1=w1+learning_rate*error*x1
        w2=w2+learning_rate*error*x2
        bias=bias+learning_rate*error

        print("input:",input[i])
        print("output:",output)
        print("error:",error )


        

```

```python
print("final W1:",w1)
print("final w2:",w2)
print("final bias:",bias)

print("\nTesting")

for x in input:
    net = x[0] * w1 + x[1] * w2 + bias

    if net >= 0:
        print(x, "->", 1)
    else:
        print(x, "->", 0)
```
